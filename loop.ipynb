```python
l=[1,2,3,4,4,"sudh","kumar",324,34.567,"abs"]
```

```python
l1=[]
l2=[]
for i in l:
  if type(i)==int or type(i)==float:
    l1.append(i)
  else:
    l2.append(i)
print(l1)
print(l2)
print("end")
```

```python
li=[4,2,True,False,"sudh",34.456]
```

```python
l1=[]
l2=[]
l3=[]
l4=[]
for i in li:
  if type(i)==int:
    l1.append(i)
  elif type(i)==float:
    l2.append(i)
  elif type(i)==str:
    l3.append(i)
  else:
    l4.append(i)
print(l1)
print(l2)
print(l3)
print(l4)
print("end")
```
